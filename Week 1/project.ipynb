## Part 7: Build Efficient Frontier Portfolios

Compute 101 portfolios along the mean-variance efficient frontier with evenly spaced estimated returns.
The first portfolio should be the minimum risk portfolio from part 4, and the last portfolio should be the maximum return portfolio from part 3.
The estimated return of each portfolio should be higher than the previous by one percent of the difference between the first and last portfolios.
That is, the estimated return of the portfolios should be similar to `np.linspace(min_risk_return, max_return, 101)`.


```python
# so this sounds like computing a "target return" across a range
# here is proof of concept getting a target return

target_return = 0.04
x_target_return = cp.Variable(n)
objective_target_return = cp.Minimize(x_target_return.T @ risk_df.to_numpy() @ x_target_return)
prob_target_return = cp.Problem(objective_target_return, [
        0 <= x_target_return, 
        cp.sum(x_target_return) == 1,
        mean_returns.to_numpy().reshape(1, -1) @ x_target_return == target_return
    ]
)
covariance_target_return = prob_target_return.solve()
covariance_target_return
```

```python
estimated_return_target_return = x_target_return.value @ mean_returns
estimated_return_target_return
```

```python
x_target_return.value.round(4)
```

```python
import numpy as np


estimated_variances = monthly_returns.drop(columns='date').var()
plt.plot(np.sqrt(estimated_variances), mean_returns, label='assets', linestyle='none', marker='o')


plt.plot(np.sqrt(covariance_minimum_risk), estimated_return_minimum_risk, label='minimum risk', marker='x', linestyle='none')
plt.plot(np.sqrt(covariance_target_return), estimated_return_target_return, label=f'target return={target_return:.2f}', marker='x', linestyle='none')
plt.legend()
plt.xlim(0)
plt.ylim(0)
plt.xlabel("expected return")
plt.ylabel("standard deviation")
plt.title("Comparison of Minimum Risk and Target Return Portfolios")
None
```

```python
target_returns = np.linspace(estimated_return_minimum_risk, estimated_return_maximum_return, 101)
target_returns[:5], target_returns[-5:]
```

```python
efficient_frontier_dfs = []
for i, target_return in enumerate(target_returns):
    x_target_return = cp.Variable(n)
    objective_target_return = cp.Minimize(x_target_return.T @ risk_df.to_numpy() @ x_target_return)
    prob_target_return = cp.Problem(objective_target_return, [
            0 <= x_target_return, 
            cp.sum(x_target_return) == 1,
            mean_returns.to_numpy().reshape(1, -1) @ x_target_return == target_return
        ]
    )
    covariance_target_return = prob_target_return.solve()
    
    risk = np.sqrt(covariance_target_return)

    efficient_frontier_df = pd.DataFrame({
        'asset': mean_returns.index,
        'allocation': x_target_return.value
    })
    efficient_frontier_df = efficient_frontier_df.set_index('asset').T.reset_index(drop=True)
    efficient_frontier_df['return'] = x_target_return.value @ mean_returns
    efficient_frontier_df['risk'] = risk

    efficient_frontier_dfs.append(efficient_frontier_df)
result = pd.concat(efficient_frontier_dfs)
result.reset_index(inplace=True, drop=True)
result
```

## Part 8: Check Maximum Return Portfolio Stability

Check the stability of the maximum return portfolio by resampling the estimated risk/return model.

Repeat 1000 times -
1. Use `np.random.multivariate_normal` to generate 23 return samples using your previously estimated risks and returns.
2. Estimate the return of each asset using that resampled return history.
3. Check which asset had the highest return in those resampled estimates.

This procedure is a reduced and simplified version of the Michaud resampled efficient frontier procedure that takes uncertainty in the risk model into account.

```python
mean_returns
```

```python
risk_df # covariance matrix
```

```python
n_repeats = 1_000
asset_names = mean_returns.index
winner_counts = np.zeros(len(asset_names), dtype=int)

for repeat in range(n_repeats):
    # Check stability of max return portfolio by resampling the estimated risk/return model

    # Generate 23 return samples using previously estimated risks and rewards
    samples = np.random.multivariate_normal(mean=mean_returns, cov=risk_df, size=23)

    # Estimate return of each asset using that resampled return history
    resampled_mean_returns = samples.mean(axis=0)

    # Check which asset had highest return in those resampled estimates
    winning_asset_index = np.argmax(resampled_mean_returns)
    winner_counts[winning_asset_index] += 1
winner_counts
```

Save a file "max_return_probabilities.tsv" with the distribution of highest return assets.
The header row should have columns "asset" and "probability".
There should be a data row for each asset and its sample probability of having the highest return based on those 1000 resampled estimates.


```python
max_return_probabilities_df = pd.DataFrame({
    'asset': asset_names,
    'probability': winner_counts/n_repeats
})
max_return_probabilities_df
```
